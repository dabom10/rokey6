<신경망 스타일 전이style transfer

**이제까지는 분류기만 바꾸는 전이학습을 했다면, 이제부터는 신경망의 일부만 가져오는 전이학습을 할 것 Ex. 저수준-중수준-고수준에서 중수준까지만 보겠다**
* VGG 특징맵을 고정 특징추출기로 구성하고,
* styleCan: 실제 이미지->캐릭터화

- NST(neural style transfer) : 이미지에서 스타일/콘텐츠 분리, 하나의 이미지에 다른 이미지의 스타일을 적용하는 딥러닝 기술.
- 콘텐츠 이미지 : 최종 이미지가 표현하려는 실제 장면이나 형태를 가진 이미지
  - content representation : 흰색 노이즈 이미지에 대해 경사하강법 수행해서 원본 이미지의 특징과 일치하는 이미지 찾음
- 스타일 이미지 Style representation : 예술 작품이나 그림에서 사용하는 특정 스타일을 가진 이미지. 질감 색상 패턴 등 시각적 표현 특징 유지위해 사용->여러 가중치. Gram Matrix(특징 맵 간 상관관계) 비산 (ex.청바지-정장 nope!)

- Gram Matrix : 상관관계 행렬(특징 벡터간 내적으로 만들어진다) "이 이미지의 어떤 시각적 특징들이 얼마나 자주 함께 나타나는가?"

- 단점: 고해상도 이미지 처리시 시간,메모리 많이 소요/세밀한 스타일 변화는 콘텐츠 왜곡 위험이 있음/스타일 손실을 너무 강조하면 콘텐츠 의미가 손상될 수 있어 균형 중요
