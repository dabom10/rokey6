
1. CNN의 특징
- 가중치 공유: 파라미터 수 엄청 줄어듦(기존 신경망ex.DNN: 픽셀 하나하나 다른 가중치 줌)
- 자동 특징추출: 패턴 학습
- 여러 은닉층 구조: 저~고수준 특징
- AlexNet - VGG - ...

2. LeNet-5 : 2개 합성곱(CNN) 계층[feature extractor] + 3개의 완전연결(FCN) 계층
*subsampling으로 이미지 공간 줄임

3. AlexNet : LeNet 보완-> 8개 계층: 5개의 Feature extractor + 3개의 classfier(FCN), maxfooling, ImageNet

5. VGGNet : 13개 계층(10개의 합성곱 계층, 3개의 완전연결 계층): 계산량 엄청 많음(GPU 써야함), AlexNet 위에 계층을 더 깊게 쌓고, 합성곱 ㄱ커널은 더 작게 해서 깊이에 집중
VGG13,16,19 중에 16,19 많이씀, VGG16_bn(bn계층추가) 모델도 있다

6. CNN 기반 분류 앱 - Google LeNet, Inception 모듈

- Inception 모듈 : 순서대로 필터 돌리지 말고 한번에 돌리면 안돼? 병렬 합성곱 계층, 1*1 합성곱, FC 대신 전역 평균 풀링 사용, 22개 계층을 쓰지만 파라미터는 VGG보다 훨씬 작게 씀

7. Anchor-based VS Anchor-free
- Anchor-based: 앵커박스 크기로부터 객체 크기에 가까워지도록 클러스터링 알고리즘으로 위치에 대한 scale 값 추론, 느림, YOLOv2-7
- Anchor-free: 특징맵으로 구성된 그리드 영역으로부터 객체에 얼마나 떨어져 있냐=거리로 scale 추론, YOLOv8 ~
