OpenCV Optical Flow & Object Tracking 정리

1. 템플릿 매칭 (Template Matching)

- 템플릿 이미지와 원본 이미지를 직접 비교하여 위치 탐색
- cv2.matchTemplate()
- 단점: 밝기, 회전, 색 변화에 민감

2. 특징자 기반 추적 (Feature-based Tracking)
- 특징자 (descriptor): 모서리, 점 등 눈에 띄는 강한 부분
- 특징자 검출 (Feature Detection): 이미지에서 특징점을 자동으로 찾아 기준점으로 사용

3. MeanShift 추적
- ROI 색 히스토그램 확률분포 기반 이동
- 빠르고 단순함
- 제한: 크기·회전 변화에 적응 불가, 박스 크기 고정

4. CamShift (Continuously Adaptive MeanShift)
- MeanShift 개선 **캠휴hue를 따라간다**
- 매 프레임마다 윈도우 크기와 방향 자동 갱신
- 크기 변화·회전에 적응 가능

5. 루카스-카나데 Optical Flow (Lucas-Kanade, LK)
- 두 프레임 사이 특징점(sparse) 이동 추적
- 가정: 밝기 보존(물체 한 점은 밝기 유지)/작은 이동(프레임 다음으로 갈때 점 조금만 이동)/국소 영역(인접한 픽셀은 비슷한 움직임)
- 빠르고 안정적

6. Dense Optical Flow
- 모든 픽셀에 대한 움직임 계산
- 예: Farneback, TV-L1
- RAFT (Recurrent All-Pairs Field Transform)
  - 딥러닝 기반 최고 수준 optical flow
  - 모든 픽셀 쌍의 상관관계 계산
  - 반복적으로 flow를 정제해 정확한 결과 생성(recurrent 구조)

**전체 구조 정리**
```
객체 추적 (Object Tracking)
│
├── 1) 템플릿 기반
│    └── Template Matching
│
├── 2) 색 분포 기반
│    ├── MeanShift
│    └── CamShift
│
├── 3) 특징점 기반 (Sparse Optical Flow)
│    └── Lucas-Kanade
│
└── 4) Dense Optical Flow
     ├── Farneback, TV-L1
     └── RAFT (딥러닝)


```
객체 추적 (Object Tracking)
│
├── 1) 픽셀 패턴 기반 (Template / Matching 계열)
│     └── Template Matching
│          - 픽셀 그대로 비교
│          - 밝기/회전 변화에 약함
│
├── 2) 색 히스토그램 기반 (MeanShift 계열)
│     ├── MeanShift
│     │     - 색 분포(히스토그램) 중심으로 이동
│     │     - 크기/회전 변화 X
│     └── CamShift
│           - MeanShift 업그레이드
│           - 크기/회전 자동 적응
│
├── 3) 특징점 기반 (Sparse Optical Flow 계열)
│     └── Lucas-Kanade (LK Optical Flow) 루카스파스
│           - 특징점들의 이동만 추적
│           - 빠르고 안정적
│
└── 4) 픽셀 전체 기반 (Dense Optical Flow 계열)
      ├── Farneback, TV-L1
      └── RAFT (딥러닝 SOTA)
            - 모든 픽셀 쌍 상관관계 계산
            - 매우 정확한 flow 추정
